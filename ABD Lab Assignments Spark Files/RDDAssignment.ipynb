# Q1. Create RDDs in three different ways

# Q3. Find the frequency of each word using RDD

```python
# Read the file
rdd = sc.textFile("input.txt")

# Split lines into words
words = rdd.flatMap(lambda line: line.split())

# Give each word a count of 1
pairs = words.map(lambda word: (word, 1))

# Add counts of the same words
result = pairs.reduceByKey(lambda a, b: a + b)

# Display the result
print(result.collect())
```

# Q10. Map the Registration numbers to corresponding branch. 58000 series BDA, 57000 series AIML, 38000 series VLSI, 39000 series ES, and 47000 series CDC. Given registration number, generate a key-value pair of Registration Number and Corresponding Branch.

```python
# Create registration numbers
numbers = [58001, 57002, 38003, 39004, 47005, 58006]

# Create RDD
rdd = sc.parallelize(numbers)

# Map registration number to branch
def get_branch(num):
    if num // 1000 == 58:
        return (num, "BDA")
    elif num // 1000 == 57:
        return (num, "AIML")
    elif num // 1000 == 38:
        return (num, "VLSI")
    elif num // 1000 == 39:
        return (num, "ES")
    elif num // 1000 == 47:
        return (num, "CDC")
    else:
        return (num, "Unknown")

result = rdd.map(get_branch)

print(result.collect())
```

# 11.	Text file contain numbers. Numbers are separated by one white space. There is no order to store the numbers. One line may contain one or more numbers. Find the maximum, minimum, sum and mean of numbers.

3.	Find the maximum and minimum temperature for city Bangalore from the given dataset.


```python
# Read dataset
rdd = sc.textFile("temperature.csv")

# Remove header
data = rdd.filter(lambda line: not line.startswith("Date"))

# Select Bangalore records
city_data = data.filter(
    lambda line: line.split(",")[2].strip().lower() == "bangalore"
)

# Extract temperatures
temps = city_data.map(lambda line: float(line.split(",")[1]))

# Display results
if not temps.isEmpty():
    print("Maximum temperature:", temps.max())
    print("Minimum temperature:", temps.min())
else:
    print("Bangalore data not found")
```

# Create dataset (text file) of bank transactions. Fields in file are ‘Bank ID’, ‘Account Number’, ‘Transaction Date’, ‘Transaction Type’ (credit or debit), ‘Transaction Amount’. Date format is dd-mm-yyyy.
1.	Count unique number of customers
2.	Count unique number of Bank ID

3.	Count unique number of customers per Bank ID

4.	Number of transactions for given Account Number
5.	Number of credit transactions for given Account Number in a given year


```python
with open("bank.txt", "w") as f:
    f.write("B101,A1001,12-01-2024,credit,5000\n")
    f.write("B101,A1002,15-02-2024,debit,1200\n")
    f.write("B102,A1003,20-03-2024,credit,7000\n")
    f.write("B101,A1001,10-04-2024,debit,500\n")
    f.write("B102,A1004,05-06-2023,credit,3000\n")
    f.write("B103,A1005,11-07-2024,credit,9000\n")
    f.write("B102,A1003,18-08-2024,debit,1500\n")
    f.write("B103,A1006,21-09-2023,credit,4000\n")
    f.write("B101,A1002,12-10-2024,credit,2000\n")
    f.write("B103,A1005,25-11-2024,debit,1000\n")
    f.write("B101,A1001,12-10-2024,credit,50000\n")
```

```python
from pyspark import SparkContext

sc = SparkContext.getOrCreate()

rdd = sc.textFile("bank.txt")

# Split each line into fields
data = rdd.map(lambda line: line.split(","))

# Fields:
# 0 = Bank ID
# 1 = Account Number
# 2 = Transaction Date
# 3 = Transaction Type
# 4 = Transaction Amount
```

1.	Count unique number of customers


```python
customers = data.map(lambda x: x[1]).distinct()

print("Unique customers:", customers.count())
```

2.	Count unique number of Bank ID



```python
banks = data.map(lambda x: x[0]).distinct()

print("Unique Bank IDs:", banks.count())
```

3.	Count unique number of customers per Bank ID



```python
result = data.map(lambda x: (x[0], x[1])) \
             .distinct() \
             .map(lambda x: (x[0], 1)) \
             .reduceByKey(lambda a, b: a + b)

print(result.collect())
```

4.	Number of transactions for given Account Number



```python
account = "A1001"

result = data.filter(lambda x: x[1] == account).count()

print("Number of transactions:", result)
```

5.	Number of credit transactions for given Account Number in a given year

```python
account = "A1001"
year = "2024"

result = data.filter(
    lambda x: x[1] == account
    and x[3].lower() == "credit"
    and x[2].split("-")[2] == year
).count()

print("Credit transactions:", result)
```
